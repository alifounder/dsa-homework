## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
#1
def to_num(s):
    s = s.replace(" ", "")
    return float(s)
    
def clean_city(s):
    s = s.strip().lower()
    names = {"almaty": "алматы", "astana": "астана", "shymkent": "шымкент"}
    if s in names:
        s = names[s]
    return s.capitalize()

print(to_num("5 555"), to_num("2026.0"))
print(clean_city(" Almaty"), clean_city("астана"), clean_city("Shymkent"))
````

````python
#2
cnt = {}
for r in rows:
    cat = r["category"]
    cnt[cat] = cnt.get(cat, 0) + 1

res = sorted(cnt.items(), key=lambda x: x[1], reverse=True)
for cat, n in res:
    print(cat, n)
````

````python
#3
total = {}
num = {}
for r in rows:
    if r["status"] == "оплачен":
        cat = r["category"]
        total[cat] = total.get(cat, 0) + to_num(r["amount"])
        num[cat] = num.get(cat, 0) + 1

print("категория | выручка | средний чек")
for cat in total:
    avg = total[cat] / num[cat]
    print(f"{cat} | {total[cat]:.0f} | {avg:.0f}")
````

````python
#4
months = {}
for r in rows:
    m = r["order_date"][:7]
    months[m] = months.get(m, 0) + 1

for m in sorted(months):
    print(m, months[m])
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отмененных заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
#5
rev = {}
for r in rows:
    if r["status"] == "оплачен":
        m = r["order_date"][:7]
        rev[m] = rev.get(m, 0) + to_num(r["amount"])

for m in sorted(rev):
    print(m, rev[m])

change = (rev["2026-10"] - rev["2026-09"]) / rev["2026-09"] * 100
print(f"Изменение: {change:.1f}%")
````

````python
#6
all_c = {}
canc = {}
for r in rows:
    c = r["city"]
    all_c[c] = all_c.get(c, 0) + 1
    if r["status"] == "отменён":
        canc[c] = canc.get(c, 0) + 1

for c in all_c:
    p = canc.get(c, 0) / all_c[c] * 100
    print(f"{c}: {p:.1f}%")
````

````python
#7
cl = {}
for r in rows:
    if r["status"] == "оплачен":
        k = r["customer_id"]
        cl[k] = cl.get(k, 0) + to_num(r["amount"])

top = sorted(cl.items(), key=lambda x: x[1], reverse=True)[:5]
for k, v in top:
    print(k, v)
````

На первом месте клиент TEST. Судя по сумме и названию это не настоящий клиент, а тестовый. У него всего 3 заказа, все по 9999990 и все в категории электроника. Его выручка почти 30 млн, это в сто с лишним раз больше, чем у второго места.

````python
#8
cust = {}
for r in rows:
    c = r["city"]
    if c not in cust:
        cust[c] = set()
    cust[c].add(r["customer_id"])

for c in cust:
    print(c, len(cust[c]))
````
